# KPI Engineering

## Higher Education Performance Dashboard

### Objective
This notebook generates the Key Performance Indicators (KPIs) required for the Higher Education Performance Dashboard.

The KPIs generated are:

- Global Ranking Score+
- Research Impact Score
- Faculty-to-Student Ratio
- International Student Percentage
- Academic Reputation Score
- Research Productivity Index

## Step 1: Import Required Libraries
Import the required Python libraries for data processing and KPI generation

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

## Step 2: Load the Final Enriched Dataset

Load the processed university dataset that contains cleaned and enriched university information. This dataset serves as the input for KPI engineering.

```python
path = "../datasets/final_dataset/university_final_dataset.xlsx"

df = pd.read_excel(path)

print(df.shape)
df.head()
```

## Step 3: Validate Required Columns

Verify that all necessary columns required for KPI calculation are available in the dataset before performing any computations.

```python
required_columns = [
    "world_rank",
    "academic_reputation_score",
    "citations_count",
    "Research_Output_x",
    "research_productivity_index",
    "faculty_count",
    "student_population",
    "international_student_percentage"
]

missing = [c for c in required_columns if c not in df.columns]

if missing:
    print("Missing Columns:", missing)
else:
    print("All Required Columns Found.")
```

## Step 4: KPI 1 – Global Ranking Score

The Global Ranking Score provides a standardized ranking metric where a higher score represents better university performance.

Formula:

Global Ranking Score = 1000 − World Rank

```python
df["global_ranking_score"] = (
    1000 - pd.to_numeric(df["world_rank"], errors="coerce")
)
```

## Step 5: KPI 2 – Research Impact Score

The Research Impact Score evaluates the overall research performance of a university by combining citation performance, research output, and research productivity.

Formula:

Research Impact Score = Average of
- Citations Count
- Research Output
- Research Productivity Index

```python
df["research_impact_score"] = (
    df[
        [
            "citations_count",
            "Research_Output_x",
            "research_productivity_index"
        ]
    ]
    .mean(axis=1)
)
```

```python
print(df[["student_population", "faculty_count", "international_students_count"]].isna().sum())
print((df[["student_population", "faculty_count", "international_students_count"]] == -1).sum())
```

## Step 6: KPI 3 – Faculty-to-Student Ratio

The Faculty-to-Student Ratio measures the academic support available to students.

Formula:

Faculty-to-Student Ratio = Student Population ÷ Faculty Count

Lower ratios generally indicate better academic support.

```python
mask = (
    df["faculty_count"].notna()
    &
    df["student_population"].notna()
    &
    (df["faculty_count"] > 0)
    &
    (df["student_population"] > 0)
)

df.loc[mask, "faculty_to_student_ratio"] = (
    df.loc[mask, "student_population"]
    / df.loc[mask, "faculty_count"]
)
```

```python
df[
    [
        "international_students_count",
        "student_population",
        "international_student_percentage"
    ]
].head(10)
```

## Step 7: KPI 4 – International Student Percentage

This KPI measures the proportion of international students enrolled in a university.

Formula:

International Student Percentage =
(International Students ÷ Student Population) × 100

```python
# KPI 4
# International Student Percentage

df["international_student_percentage"] = pd.to_numeric(
    df["international_student_percentage"],
    errors="coerce"
)

print("International Student Percentage KPI validated successfully.")
```

## Step 8: KPI 5 – Academic Reputation Score

The Academic Reputation Score reflects the global academic reputation of a university based on ranking data.

The score is standardized and stored as a numerical KPI for dashboard analysis.

```python
df["academic_reputation_score"] = (
    pd.to_numeric(
        df["academic_reputation_score"],
        errors="coerce"
    )
)
```

## Step 9: KPI 6 – Research Productivity Index

The Research Productivity Index measures the research efficiency of a university.

This KPI is used to compare institutional research performance across universities.

```python
df["research_productivity_index"] = (
    pd.to_numeric(
        df["research_productivity_index"],
        errors="coerce"
    )
)
```

## Step 10: KPI Summary Statistics

Display descriptive statistics for all generated KPIs to verify their distributions and ensure successful KPI generation.

```python
kpis = [
    "global_ranking_score",
    "research_impact_score",
    "faculty_to_student_ratio",
    "international_student_percentage",
    "academic_reputation_score",
    "research_productivity_index"
]

df[kpis].describe()
```

## Step 11: Save the Final Dataset

Save the dataset with all generated KPI columns as the final project dataset.

Output File:

`university_final_dataset.xlsx`

```python
output_path = "../datasets/final_dataset/university_final_dataset.xlsx"

df.to_excel(
    output_path,
    index=False
)

print("=" * 60)
print("KPI Engineering Completed Successfully")
print("Final Dataset Shape:", df.shape)
print("Dataset Saved Successfully")
print("=" * 60)
```
